```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.spatial import distance
from scipy.stats import entropy
from sklearn.metrics import confusion_matrix, classification_report, mean_absolute_error, roc_auc_score, f1_score
import random

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.applications import ResNet50
from tensorflow.keras.applications.resnet50 import preprocess_input

# ============================================================
# 1) CONFIG, SEED & HARD LABEL TANIMLARI
# ============================================================
# --- TEKRARLANABİLİRLİK İÇİN SEED AYARI ---
SEED = 534
np.random.seed(SEED)
tf.random.set_seed(SEED)
random.seed(SEED)

DATA_ROOT = "/content/drive/MyDrive/PhD/3-Colab_for_classification/data/_TRAIN_multi_input_soft_classes_"
SAVE_PATH = f"/content/drive/MyDrive/PhD/3-Colab_for_classification/models/permutation_clean/clean_model_{SEED}.keras"
EXCEL_PATH = f"/content/drive/MyDrive/PhD/3-Colab_for_classification/models/permutation_clean/clean_model_{SEED}.xlsx"

IMG_SIZE = (224, 224)
BATCH_SIZE = 4
VAL_SPLIT = 0.2
TEST_SPLIT = 0.1

# 🌟 BAYESIAN OPTIMIZATION ILE BULUNAN PARAMETRELER
BEST_DENSE_1 = 1024
BEST_DENSE_2 = 384
BEST_DROPOUT_1 = 0.30
BEST_DROPOUT_2 = 0.20
BEST_LR_STAGE1 = 3.36e-05
BEST_LR_STAGE2 = 3.88e-05
BEST_FT_LAYERS = 20

CLASS_MAP = {
    "air_layer": 0,
    "bubbly": 1,
    "transition": 2,
    "b_to_t": 2
}
TARGET_NAMES_3 = ["Air", "Bubbly", "Transition"]

# ============================================================
# 2) VERİ YÜKLEME, BÖLME VE SLIDING WINDOW / ROLL ÇOĞALTMA
# ============================================================

def get_base_groups(root_dir):
    """
    Sadece orijinal klasör gruplarını toplar.
    Henüz çoğaltma (sliding/rolling) işlemini yapmaz.
    """
    group_data = [] # (klasördeki_tüm_dosya_yolları, etiket)
    class_folders = sorted([d for d in os.listdir(root_dir) if os.path.isdir(os.path.join(root_dir, d))])

    for folder in class_folders:
        label_idx = CLASS_MAP.get(folder)
        if label_idx is None: continue

        cls_path = os.path.join(root_dir, folder)
        group_dirs = sorted([os.path.join(cls_path, d) for d in os.listdir(cls_path)])
        for g_dir in group_dirs:
            files = sorted([os.path.join(g_dir, f) for f in os.listdir(g_dir)
                            if f.lower().endswith(('.png', '.jpg', '.jpeg', '.tif'))])

            # 4-input modeli için klasörde en az 4 resim olmalıdır
            if len(files) >= 4:
                group_data.append((files, label_idx))

    return group_data

# 1. Adım: Temel grupları çekelim
base_groups = get_base_groups(DATA_ROOT)

# 2. Adım: Grupları karıştırıp bölüyoruz (Veri sızıntısını önleyen en kritik adım)
random.seed(SEED)
random.shuffle(base_groups)

total_len = len(base_groups)
test_end = int(total_len * TEST_SPLIT)
val_end = test_end + int(total_len * VAL_SPLIT)

test_groups = base_groups[:test_end]
val_groups = base_groups[test_end:val_end]
train_groups = base_groups[val_end:]

def expand_with_sliding_window(groups):
    """
    Her grubu kendi içinde kaydırarak veri sayısını artırır.
    Klasörde tam 4 resim varsa dairesel döndürür (Cyclic Roll),
    daha fazla resim varsa pencereler halinde kaydırır (Sliding Window).
    """
    expanded_paths = []
    expanded_labels = []

    for files, label in groups:
        n = len(files)

        if n == 4:
            # --- Senaryo A: Tam 4 resim varsa Cyclic Roll (4 katı veri) ---
            for shift in range(4):
                rolled = files[shift:] + files[:shift]
                expanded_paths.append(rolled)
                expanded_labels.append(label)
        else:
            # --- Senaryo B: 4'ten fazla resim varsa Sliding Window ---
            for i in range(n - 3):
                window = files[i:i+4]
                expanded_paths.append(window)
                expanded_labels.append(label)

    return np.array(expanded_paths), np.array(expanded_labels)

# 3. Adım: Setleri birbirine karıştırmadan, bağımsız olarak çoğaltıyoruz
train_paths, train_labels = expand_with_sliding_window(train_groups)
val_paths, val_labels = expand_with_sliding_window(val_groups)
test_paths, test_labels = expand_with_sliding_window(test_groups)

print(f"📈 Orijinal Grup Sayıları: {len(train_groups)} Train | {len(val_groups)} Val | {len(test_groups)} Test")
print(f"🚀 Çoğaltılmış Set Boyutları: {len(train_paths)} Train | {len(val_paths)} Val | {len(test_paths)} Test")

# --- İYİLEŞTİRİLMİŞ VERİ İŞLEME FONKSİYONLARI ---

# YALNIZCA EĞİTİM SETİ İÇİN: 4 resme de AYNI ve DAHA ÇEŞİTLİ augmentation'ı uygular
def process_quad_with_sync_aug(paths, label):
    imgs = []
    for i in range(4):
        img = tf.io.read_file(paths[i])
        img = tf.image.decode_image(img, channels=3, expand_animations=False)
        img = tf.image.convert_image_dtype(img, tf.float32)
        img = tf.image.resize(img, [int(IMG_SIZE[0]*1.1), int(IMG_SIZE[1]*1.1)]) # Kırpma payı
        imgs.append(img)

    main_seed = tf.random.uniform([2], maxval=tf.int32.max, dtype=tf.int32)

    # --- 1. Senkronize Rastgele Kırpma ve Yakınlaştırma ---
    bbox_seed = tf.random.experimental.stateless_split(main_seed, num=1)[0, :]
    shape = tf.shape(imgs[0])
    begin, size, _ = tf.image.stateless_sample_distorted_bounding_box(
        shape,
        bounding_boxes=tf.constant([0.0, 0.0, 1.0, 1.0], dtype=tf.float32, shape=[1, 1, 4]),
        seed=bbox_seed,
        min_object_covered=0.75,
        aspect_ratio_range=(0.8, 1.2),
        area_range=(0.8, 1.0),
        max_attempts=100
    )

    augmented_imgs = []
    for img in imgs:
        aug_img = tf.slice(img, begin, size)
        aug_img = tf.image.resize(aug_img, IMG_SIZE)

        # --- 2. Renk Değişimleri ---
        color_seed = tf.random.experimental.stateless_split(main_seed, num=1)[0, :]
        aug_img = tf.image.stateless_random_brightness(aug_img, max_delta=0.1, seed=color_seed)
        aug_img = tf.image.stateless_random_contrast(aug_img, lower=0.9, upper=1.1, seed=color_seed)
        aug_img = tf.image.stateless_random_saturation(aug_img, lower=0.9, upper=1.1, seed=color_seed)

        aug_img = tf.clip_by_value(aug_img, 0.0, 1.0)

        # Ön işleme
        aug_img = preprocess_input(aug_img * 255.0)
        augmented_imgs.append(aug_img)

    return (augmented_imgs[0], augmented_imgs[1], augmented_imgs[2], augmented_imgs[3]), label

# VALIDATION VE TEST SETİ İÇİN: Augmentation uygulanmaz
def process_quad_no_aug(paths, label):
    imgs = []
    for i in range(4):
        img = tf.io.read_file(paths[i])
        img = tf.image.decode_image(img, channels=3, expand_animations=False)
        img = tf.image.resize(img, IMG_SIZE)
        img = preprocess_input(img)
        imgs.append(img)
    return (imgs[0], imgs[1], imgs[2], imgs[3]), label

# --- OPTİMİZE EDİLMİŞ TF DATASET PIPELINE ---
AUTOTUNE = tf.data.AUTOTUNE

train_ds = tf.data.Dataset.from_tensor_slices((train_paths, train_labels)) \
    .shuffle(len(train_paths), seed=SEED) \
    .map(process_quad_with_sync_aug, num_parallel_calls=AUTOTUNE) \
    .batch(BATCH_SIZE) \
    .prefetch(AUTOTUNE) \
    .cache()

val_ds = tf.data.Dataset.from_tensor_slices((val_paths, val_labels)) \
    .map(process_quad_no_aug, num_parallel_calls=AUTOTUNE) \
    .batch(BATCH_SIZE) \
    .prefetch(AUTOTUNE) \
    .cache()

test_ds = tf.data.Dataset.from_tensor_slices((test_paths, test_labels)) \
    .map(process_quad_no_aug, num_parallel_calls=AUTOTUNE) \
    .batch(BATCH_SIZE) \
    .prefetch(AUTOTUNE) \
    .cache()

print(f"✅ TF Veri Akış Hatları Hazır!")

# ============================================================
# 3) MODEL MİMARİSİ
# ============================================================
base_model = ResNet50(weights="imagenet", include_top=False, input_shape=(224, 224, 3), name="resnet50")
base_model.trainable = False

inputs = [layers.Input(shape=(224, 224, 3), name=f"input_{i+1}") for i in range(4)]
features = []
for inp in inputs:
    x = base_model(inp, training=False)
    x = layers.GlobalAveragePooling2D()(x)
    features.append(x)

combined = layers.Concatenate()(features)

# Dense katmanlarında Bayesian Optimization hiperparametreleri kullanılıyor
x = layers.Dense(BEST_DENSE_1, activation="relu")(combined)
x = layers.BatchNormalization()(x)
x = layers.Dropout(BEST_DROPOUT_1)(x)
x = layers.Dense(BEST_DENSE_2, activation="relu")(x)
x = layers.Dropout(BEST_DROPOUT_2)(x)
outputs = layers.Dense(3, activation="softmax")(x)

model = keras.Model(inputs=inputs, outputs=outputs)
model.summary()

# ============================================================
# 4) MODEL EĞİTİMİ (STAGE 1 & STAGE 2)
# ============================================================
checkpoint = keras.callbacks.ModelCheckpoint(SAVE_PATH, monitor='val_loss', save_best_only=True, verbose=1)
reduce_lr = keras.callbacks.ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=5, verbose=1)
early_stop = keras.callbacks.EarlyStopping(monitor='val_loss', patience=20, restore_best_weights=True, verbose=1)

print("\n🚀 Aşama 1: Frozen Training...")
model.compile(optimizer=keras.optimizers.Adam(BEST_LR_STAGE1), loss="sparse_categorical_crossentropy", metrics=["accuracy"])
h1 = model.fit(train_ds, validation_data=val_ds, epochs=20, callbacks=[checkpoint, reduce_lr])

print("\n✅ En iyi model ağırlıkları yükleniyor. Fine-tuning başlıyor...")
model = keras.models.load_model(SAVE_PATH)

print("\n🔥 Aşama 2: Fine-Tuning...")
base_model_loaded = model.get_layer("resnet50")
base_model_loaded.trainable = True

# Seçilen fine-tune katman sayısı kadar sondan geriye katman açılır
for layer in base_model_loaded.layers[:-BEST_FT_LAYERS]:
    layer.trainable = False

model.compile(optimizer=keras.optimizers.Adam(BEST_LR_STAGE2), loss="sparse_categorical_crossentropy", metrics=["accuracy"])
h2 = model.fit(train_ds, validation_data=val_ds, epochs=100, callbacks=[checkpoint, reduce_lr, early_stop])

full_history = {key: h1.history[key] + h2.history[key] for key in h1.history.keys()}

# ============================================================
# 5) GELİŞMİŞ DETAYLI ANALİZ
# ============================================================
def calculate_comprehensive_metrics(dataset, dataset_name="SET"):
    print(f"\n" + "="*60)
    print(f"📊 {dataset_name} DETAYLI PERFORMANS ANALİZİ")
    print("="*60)

    y_true_idx, y_prob_pred = [], []
    for x_batch, y_batch in dataset:
        preds = model.predict(x_batch, verbose=0)
        y_prob_pred.extend(preds)
        y_true_idx.extend(y_batch.numpy())

    y_true_idx = np.array(y_true_idx)
    y_prob_pred = np.array(y_prob_pred)
    y_pred_idx = np.argmax(y_prob_pred, axis=1)
    y_true_soft_converted = np.eye(len(TARGET_NAMES_3))[y_true_idx]

    acc = np.mean(y_true_idx == y_pred_idx)
    f1_macro = f1_score(y_true_idx, y_pred_idx, average='macro')
    mae = mean_absolute_error(y_true_soft_converted, y_prob_pred)
    try:
        auc_macro = roc_auc_score(y_true_soft_converted, y_prob_pred, multi_class='ovr', average='macro')
    except:
        auc_macro = 0.0

    eps = 1e-12
    kl_divs = [entropy(p + eps, q + eps) for p, q in zip(y_true_soft_converted, y_prob_pred)]
    js_dists = [distance.jensenshannon(p, q) for p, q in zip(y_true_soft_converted, y_prob_pred)]

    metrics_data = {
        "Metrik": ["Accuracy", "F1 Score (Macro)", "AUC (Macro)", "MAE", "K-L Divergence", "Jensen-Shannon Dist."],
        "Değer": [f"{acc:.4f}", f"{f1_macro:.4f}", f"{auc_macro:.4f}", f"{mae:.4f}", f"{np.mean(kl_divs):.4f}", f"{np.mean(js_dists):.4f}"]
    }
    print(pd.DataFrame(metrics_data).to_string(index=False))
    print("-" * 40)
    print(classification_report(y_true_idx, y_pred_idx, target_names=TARGET_NAMES_3))

    cm = confusion_matrix(y_true_idx, y_pred_idx)
    plt.figure(figsize=(6, 4))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=TARGET_NAMES_3, yticklabels=TARGET_NAMES_3)
    plt.title(f'{dataset_name} Confusion Matrix')
    plt.xlabel('Tahmin'); plt.ylabel('Gerçek'); plt.show()

    return y_true_idx, y_pred_idx, y_prob_pred

# Metriklerin hesaplanması
_, _, _ = calculate_comprehensive_metrics(train_ds, "TRAINING SET")
_, _, _ = calculate_comprehensive_metrics(val_ds, "VALIDATION SET")
y_test_true_idx, y_test_pred_idx, y_test_prob_pred = calculate_comprehensive_metrics(test_ds, "TEST SET")

# ============================================================
# 6) EXCEL RAPORLAMA (GÜNCELLENDİ)
# ============================================================
# Sızıntısız bölünmüş test_paths doğrudan kullanılıyor
report_list = []
for i in range(len(test_paths)):
    report_list.append({
        "Quad_Folder": os.path.dirname(test_paths[i][0]),
        "True_Class": TARGET_NAMES_3[y_test_true_idx[i]],
        "Predicted_Class": TARGET_NAMES_3[y_test_pred_idx[i]],
        "Status": "TRUE" if y_test_true_idx[i] == y_test_pred_idx[i] else "FALSE",
        **{f"Prob_{TARGET_NAMES_3[j]}": float(y_test_prob_pred[i][j]) for j in range(3)}
    })

os.makedirs(os.path.dirname(EXCEL_PATH), exist_ok=True)
pd.DataFrame(report_list).to_excel(EXCEL_PATH, index=False)
print(f"\n✅ Detaylı Test Raporu Kaydedildi: {EXCEL_PATH}")

# ============================================================
# 7) EĞİTİM GRAFİKLERİ
# ============================================================
def plot_final_history(history):
    plt.figure(figsize=(14, 5))
    plt.subplot(1, 2, 1)
    plt.plot(history['accuracy'], label='Train Acc', color='green')
    plt.plot(history['val_accuracy'], label='Val Acc', color='orange')
    plt.title('Accuracy vs Epoch'); plt.legend(); plt.grid(True, alpha=0.3)

    plt.subplot(1, 2, 2)
    plt.plot(history['loss'], label='Train Loss', color='red')
    plt.plot(history['val_loss'], label='Val Loss', color='blue')
    plt.title('Loss vs Epoch'); plt.legend(); plt.grid(True, alpha=0.3)
    plt.show()

plot_final_history(full_history)
```

```python
import os
import numpy as np
import pandas as pd
import random
import keras_tuner as kt

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.applications import ResNet50
from tensorflow.keras.applications.resnet50 import preprocess_input

# ============================================================
# 1) CONFIG, SEED & HARD LABEL TANIMLARI
# ============================================================
SEED = 6
np.random.seed(SEED)
tf.random.set_seed(SEED)
random.seed(SEED)

DATA_ROOT = "/content/drive/MyDrive/PhD/3-Colab_for_classification/data/_TRAIN_multi_input_soft_classes_"

IMG_SIZE = (224, 224)
BATCH_SIZE = 4
VAL_SPLIT = 0.2
TEST_SPLIT = 0.1

CLASS_MAP = {
    "air_layer": 0,
    "bubbly": 1,
    "transition": 2,
    "b_to_t": 2
}

# ============================================================
# 2) VERİ YÜKLEME, SIZINTISIZ BÖLME VE SLIDING WINDOW / ROLL ÇOĞALTMA
# ============================================================

def get_base_groups(root_dir):
    """Sadece orijinal klasör gruplarını toplar."""
    group_data = []
    class_folders = sorted([d for d in os.listdir(root_dir) if os.path.isdir(os.path.join(root_dir, d))])

    for folder in class_folders:
        label_idx = CLASS_MAP.get(folder)
        if label_idx is None: continue

        cls_path = os.path.join(root_dir, folder)
        group_dirs = sorted([os.path.join(cls_path, d) for d in os.listdir(cls_path)])
        for g_dir in group_dirs:
            files = sorted([os.path.join(g_dir, f) for f in os.listdir(g_dir)
                            if f.lower().endswith(('.png', '.jpg', '.jpeg', '.tif'))])

            if len(files) >= 4:
                group_data.append((files, label_idx))
    return group_data

# Orijinal grupları alıp karıştırarak sızıntısız bölüyoruz
base_groups = get_base_groups(DATA_ROOT)
random.seed(SEED)
random.shuffle(base_groups)

total_len = len(base_groups)
test_end = int(total_len * TEST_SPLIT)
val_end = test_end + int(total_len * VAL_SPLIT)

test_groups = base_groups[:test_end]
val_groups = base_groups[test_end:val_end]
train_groups = base_groups[val_end:]

def expand_with_sliding_window(groups):
    """Her grubu bağımsız kaydırarak dairesel veri artırımı (cyclic roll) uygular."""
    expanded_paths = []
    expanded_labels = []

    for files, label in groups:
        n = len(files)
        if n == 4:
            for shift in range(4):
                rolled = files[shift:] + files[:shift]
                expanded_paths.append(rolled)
                expanded_labels.append(label)
        else:
            for i in range(n - 3):
                window = files[i:i+4]
                expanded_paths.append(window)
                expanded_labels.append(label)

    return np.array(expanded_paths), np.array(expanded_labels)

# Çoğaltma işlemini set bazında yapıyoruz
train_paths, train_labels = expand_with_sliding_window(train_groups)
val_paths, val_labels = expand_with_sliding_window(val_groups)
test_paths, test_labels = expand_with_sliding_window(test_groups)

print(f"📈 Orijinal Grup Sayıları: {len(train_groups)} Train | {len(val_groups)} Val")
print(f"🚀 Çoğaltılmış Set Boyutları: {len(train_paths)} Train | {len(val_paths)} Val")

# --- VERİ İŞLEME FONKSİYONLARI ---
def process_quad_with_sync_aug(paths, label):
    imgs = []
    for i in range(4):
        img = tf.io.read_file(paths[i])
        img = tf.image.decode_image(img, channels=3, expand_animations=False)
        img = tf.image.convert_image_dtype(img, tf.float32)
        img = tf.image.resize(img, [int(IMG_SIZE[0]*1.1), int(IMG_SIZE[1]*1.1)])
        imgs.append(img)

    main_seed = tf.random.uniform([2], maxval=tf.int32.max, dtype=tf.int32)
    bbox_seed = tf.random.experimental.stateless_split(main_seed, num=1)[0, :]
    shape = tf.shape(imgs[0])
    begin, size, _ = tf.image.stateless_sample_distorted_bounding_box(
        shape, bounding_boxes=tf.constant([0.0, 0.0, 1.0, 1.0], dtype=tf.float32, shape=[1, 1, 4]),
        seed=bbox_seed, min_object_covered=0.75, aspect_ratio_range=(0.8, 1.2), area_range=(0.8, 1.0), max_attempts=100
    )

    augmented_imgs = []
    for img in imgs:
        aug_img = tf.slice(img, begin, size)
        aug_img = tf.image.resize(aug_img, IMG_SIZE)

        color_seed = tf.random.experimental.stateless_split(main_seed, num=1)[0, :]
        aug_img = tf.image.stateless_random_brightness(aug_img, max_delta=0.1, seed=color_seed)
        aug_img = tf.image.stateless_random_contrast(aug_img, lower=0.9, upper=1.1, seed=color_seed)
        aug_img = tf.image.stateless_random_saturation(aug_img, lower=0.9, upper=1.1, seed=color_seed)
        aug_img = tf.clip_by_value(aug_img, 0.0, 1.0)

        aug_img = preprocess_input(aug_img * 255.0)
        augmented_imgs.append(aug_img)

    return (augmented_imgs[0], augmented_imgs[1], augmented_imgs[2], augmented_imgs[3]), label

def process_quad_no_aug(paths, label):
    imgs = []
    for i in range(4):
        img = tf.io.read_file(paths[i])
        img = tf.image.decode_image(img, channels=3, expand_animations=False)
        img = tf.image.resize(img, IMG_SIZE)
        img = preprocess_input(img)
        imgs.append(img)
    return (imgs[0], imgs[1], imgs[2], imgs[3]), label

# --- DATASET PIPELINES ---
AUTOTUNE = tf.data.AUTOTUNE

train_ds = tf.data.Dataset.from_tensor_slices((train_paths, train_labels)) \
    .shuffle(len(train_paths), seed=SEED) \
    .map(process_quad_with_sync_aug, num_parallel_calls=AUTOTUNE) \
    .batch(BATCH_SIZE) \
    .prefetch(AUTOTUNE)

val_ds = tf.data.Dataset.from_tensor_slices((val_paths, val_labels)) \
    .map(process_quad_no_aug, num_parallel_calls=AUTOTUNE) \
    .batch(BATCH_SIZE) \
    .prefetch(AUTOTUNE)

# ============================================================
# 3) BAYESIAN OPTIMIZATION TUNER SINIFI
# ============================================================

class MyBayesianTuner(kt.Tuner):
    """Geriye float (val_loss) döndüren iki aşamalı eğitimleri destekleyen özel Bayesyen Tuner sınıfı."""
    def __init__(self, **kwargs):
        oracle = kt.oracles.BayesianOptimizationOracle(
            objective=kt.Objective("val_loss", direction="min"),
            max_trials=kwargs.pop("max_trials", 10),
            num_initial_points=kwargs.pop("num_initial_points", 3),
            seed=kwargs.pop("seed", 27)
        )
        super().__init__(oracle=oracle, **kwargs)

    def run_trial(self, trial):
        hp = trial.hyperparameters

        # --- Hiperparametre Uzay Tanımlamaları ---
        dense_1_units = hp.Int("dense_1_units", min_value=256, max_value=1024, step=128, default=512)
        dense_2_units = hp.Int("dense_2_units", min_value=128, max_value=512, step=64, default=256)
        dropout_1 = hp.Float("dropout_1", min_value=0.2, max_value=0.5, step=0.1, default=0.4)
        dropout_2 = hp.Float("dropout_2", min_value=0.1, max_value=0.4, step=0.1, default=0.3)
        lr_stage1 = hp.Float("lr_stage1", min_value=1e-5, max_value=1e-3, sampling="log", default=1e-4)
        lr_stage2 = hp.Float("lr_stage2", min_value=1e-6, max_value=1e-4, sampling="log", default=1e-5)
        ft_layers = hp.Int("fine_tune_layers", min_value=10, max_value=50, step=10, default=30)

        # --- Mimari Kurulumu ---
        base_model = ResNet50(weights="imagenet", include_top=False, input_shape=(224, 224, 3), name="resnet50")
        base_model.trainable = False

        inputs = [layers.Input(shape=(224, 224, 3), name=f"input_{i+1}") for i in range(4)]
        features = []
        for inp in inputs:
            x = base_model(inp, training=False)
            x = layers.GlobalAveragePooling2D()(x)
            features.append(x)

        combined = layers.Concatenate()(features)
        x = layers.Dense(dense_1_units, activation="relu")(combined)
        x = layers.BatchNormalization()(x)
        x = layers.Dropout(dropout_1)(x)
        x = layers.Dense(dense_2_units, activation="relu")(x)
        x = layers.Dropout(dropout_2)(x)
        outputs = layers.Dense(3, activation="softmax")(x)

        model = keras.Model(inputs=inputs, outputs=outputs)

        # Deneme adımı için geçici callbackler
        temp_save = f"./temp_trial_{trial.trial_id}.keras"
        checkpoint = keras.callbacks.ModelCheckpoint(temp_save, monitor='val_loss', save_best_only=True, verbose=0)
        reduce_lr = keras.callbacks.ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=3, verbose=0)
        early_stop = keras.callbacks.EarlyStopping(monitor='val_loss', patience=6, restore_best_weights=True, verbose=0)

        # --- Aşama 1: Frozen Training ---
        model.compile(optimizer=keras.optimizers.Adam(lr_stage1), loss="sparse_categorical_crossentropy", metrics=["accuracy"])
        model.fit(train_ds, validation_data=val_ds, epochs=8, callbacks=[checkpoint, reduce_lr], verbose=0)

        if os.path.exists(temp_save):
            try:
                model = keras.models.load_model(temp_save)
            except Exception:
                pass

        # --- Aşama 2: Fine-Tuning ---
        base_model_loaded = model.get_layer("resnet50")
        base_model_loaded.trainable = True
        for layer in base_model_loaded.layers[:-ft_layers]:
            layer.trainable = False

        model.compile(optimizer=keras.optimizers.Adam(lr_stage2), loss="sparse_categorical_crossentropy", metrics=["accuracy"])
        history = model.fit(train_ds, validation_data=val_ds, epochs=15, callbacks=[checkpoint, reduce_lr, early_stop], verbose=0)

        val_loss_score = min(history.history['val_loss'])

        # Temizlik
        if os.path.exists(temp_save):
            os.remove(temp_save)

        # Sonucu Oracle mekanizmasına bildiriyoruz
        self.oracle.update_trial(trial.trial_id, {"val_loss": val_loss_score})

# ============================================================
# 4) TUNER ÇALIŞTIRMA VE EN İYİ DEĞERLERİ YAZDIRMA
# ============================================================

tuner = MyBayesianTuner(
    max_trials=10,
    num_initial_points=3,
    directory="bayesian_tuning",
    project_name="resnet_quad_classification",
    seed=SEED
)

# Aramayı başlatıyoruz
print("\n🔍 Bayesian Optimization araması başlıyor...")
tuner.search()

# Arama bittiğinde en iyi parametreleri ekrana basıyoruz
print("\n🏆 OPTİMİZASYON TAMAMLANDI! EN İYİ HİPERPARAMETRELER:")
print("="*50)
best_hps = tuner.get_best_hyperparameters(num_trials=1)[0]

print(f"BEST_DENSE_1 = {best_hps.get('dense_1_units')}")
print(f"BEST_DENSE_2 = {best_hps.get('dense_2_units')}")
print(f"BEST_DROPOUT_1 = {best_hps.get('dropout_1'):.2f}")
print(f"BEST_DROPOUT_2 = {best_hps.get('dropout_2'):.2f}")
print(f"BEST_LR_STAGE1 = {best_hps.get('lr_stage1')}")
print(f"BEST_LR_STAGE2 = {best_hps.get('lr_stage2')}")
print(f"BEST_FT_LAYERS = {best_hps.get('fine_tune_layers')}")
print("="*50)
print("Bu değerleri ana eğitim kodunun başına kopyalayarak nihai eğitimi başlatabilirsin! 🚀")
```
